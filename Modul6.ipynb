## Soal 2: Metode pada kelas

Petunjuk:

* Tambahakan `self` sebagai paramater pertama pada tiap metode kelas.

JAWABAN:

````python
class Kelas:
    def __init__(self):
        self.database = {}

    def AddStudent(self, nim, name):
        self.database[nim] = name

    def RemoveStudent(self, nim):
        self.database.pop(nim)

    def Find(self, nim):
        if nim in self.database.keys():
            print('Ditemukan:', self.database[nim])
        else:
            print('Tidak ditemukan!')
    
    def Count(self):
        return len(self.database)
    

class TB4702(Kelas):
        def __init__(self):
            Kelas.__init__(self)


class TB4703(Kelas):
        def __init__(self):
            Kelas.__init__(self)


class TB4704(Kelas):
        def __init__(self):
            Kelas.__init__(self)
````

## Soal 3: Polymorphism

````python
def count(kelas): #untuk menghitung jumlah mahasiswa di kelas yang ditentukan
    kelas.Count()


def find(nim, list_kelas): # untuk mencari mahasiswa di semua kelas yang terdapat pada suatu list
    for kelas in list_kelas:
        print('Mecari di kelas:', kelas.__class__.__name__)
        kelas.Find(nim)
````
