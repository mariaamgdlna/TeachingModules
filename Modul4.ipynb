## Soal 1

````python
def enkripsi(masukan, n=3):
    masukan = masukan.lower()

    res = ''
    for huruf in masukan:
        huruf_ = ord(huruf) + n
        res = res + chr(huruf_)

    return res


def dekripsi(masukan, n=3):

    res = ''

    for huruf in masukan:
        huruf_ = ord(huruf) - n
        res = res + chr(huruf_)

    return res
````

````python
# testing

dekripsi(enkripsi("HALO! APA KABAR"))
````

## Soal 2

### Enkripsi

````python
fin = open('file.txt', 'r')

enc_lines = []
for line in fin:
    enc_line = enkripsi(line)
    enc_lines.append(enc_line)

fin.close()

fout = open('file_enc.txt', 'w')
for line in enc_lines:
    fout.write(line)

fout.close()
````

### Simpan hasil enkripsi ke file baru

````python
fin = open('file.txt', 'r')

enc_lines = []
for line in fin:
    enc_line = enkripsi(line)
    enc_lines.append(enc_line)

fin.close()

fout = open('file_enc.txt', 'w')
for line in enc_lines:
    fout.write(line)

fout.close()
````

### Dekripsi

````python
fin = open('file_enc.txt', 'r')

dec_lines = []
for line in fin:

    # [OPSIONAL] hindari \n, new line character
    if line[-1] == '\n':   
       line = line[0:-1]

    dec_line = dekripsi(line)
    dec_lines.append(dec_line)

fin.close()

for k in range(len(dec_lines)):
    print(dec_lines[k])
````
