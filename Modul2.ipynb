## Soal 1

## Soal 2

```python
bilangan = 7

flag = True

if bilangan < 2:
    flag = False

elif bilangan > 2:
    for k in range(2, bilangan):
        if bilangan % k == 0:
            flag = False
            break

if flag == False:
    print(bilangan, 'adalah BUKAN bilangan prima.')
else:
    print(bilangan, 'adalah bilangan prima.')
```

## Soal 3

```python
A = [1, -2, 0, 8, 7, 11, -1, 3]
maks = -100000

for a in A:
    if a > maks:
        maks = a

print('Nilai maksimum dari A:', maks)
```

## Soal 6

```python
f = lambda t:t**2

dx = 0.1

t = -1.0
A = 0.0

while True:
    A = A + f(t) * dx
    t = t + dx
    if t > 1:
        break

A
```
