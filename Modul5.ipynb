## SOAL 1: Membuat Kelas `Sandi`

```python
class Sandi:
    def __init__(self):
        pass


    def enkripsi(self, masukan, n=3):
        masukan = masukan.lower()

        res = ''
        for huruf in masukan:
            huruf_ = ord(huruf) + n
            res = res + chr(huruf_)

        return res


    def dekripsi(self, masukan, n=3):
        res = ''

        for huruf in masukan:
            huruf_ = ord(huruf) - n
            res = res + chr(huruf_)

        return res


    def open_file(self, fn):   
        lines = []

        fin = open(fn, 'r')    
        for line in fin:
            lines.append(line)
        fin.close()

        return lines


    def write_file(self, fn, lines):
        fout = open(fn, 'w')
        for line in lines:
            fout.write(line)
        fout.close()


    def do_encryption(self, lines, n=3):
        enc_lines = []
        for line in lines:
            enc_line = self.enkripsi(line)
            enc_lines.append(enc_line)

        return enc_lines
    

    def do_decryption(self, lines, n=3):
        dec_lines = []
        for line in lines:

            # [OPSIONAL] hindari \n, new line character
            if line[-1] == '\n':   
                line = line[0:-1]

            dec_line = self.dekripsi(line)
            dec_lines.append(dec_line)

        return dec_lines
```

## Soal 2: Program Demo  Sederhana

* Buat objek `Sandi`
* Buka file 
* Enkripsi file yang sudah dibuka
* Simpan hasilnya ke file baru
* Buka file baru tersebut
* Dekripisi isi file tersebut dan tampilkan hasilnya

```python
# Testing

# Buat objek Sandi 
S = Sandi()

# Buka file
lines = S.open_file('file.txt')
print(lines)

# Enkripsi file yang sudah dibuka
enc_lines = S.do_encryption(lines)
print(enc_lines)

# Simpan hasilnya ke file baru
S.write_file('file_enc.txt', enc_lines)

# Buka file baru tersebut
lines = S.open_file('file_enc.txt')

# Dekripisi isi file tersebut dan tampilkan hasilnya
dec_lines = S.do_decryption(lines)
print(dec_lines)
```
