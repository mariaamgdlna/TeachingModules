```python
import matplotlib.pyplot as plt
import numpy as np

np.set_printoptions(precision=4, suppress=True)
```

## Soal 1

Buatlah plot dari fungsi `y = t**2` pada rentang yang disebutkan.

**Solusi**

 Waktu (`t`) yang akan kita gunakan: `t = -1, -0.9, ..., 0.9, 1`.

```python
# siapkan vektor waktu
start = -1
stop = 1
dx = 0.1

t = np.arange(start, stop, dx) # start, stop, step
y = t**2

print(t)
print(y)
```

Gambarkan plot `y(t)`.

```python
fig, ax = plt.subplots()
ax.plot(t, y)
plt.show()
```

## Soal 2

Buatlah plot dari approximation hasil peng-integral-an `y = t**2`. (Plot $A$ sebagai fungsi waktu!).
* Tumpangkan hasilnya dengan soal no. 1. 
* Berikan warna berbeda untuk setiap plot.
* Tambahkan legend/keterangan: `x^2` dan `∫ (x^2) dx`

**Solusi**

Kumpulkan nilai `A` sebagai fungsi waktu:

```python
f = lambda t:t**2

A = 0.0

At = []
for t_ in t:
    At.append(A)
    A = A + f(t_) * dx

# Konversikan list menjadi NumPy array
At = np.array(At)
print(At)
```

```python
fig, ax = plt.subplots()
ax.plot(t, y,  color='r', label='x^2')
ax.plot(t, At, color='k', label='∫ (x^2) dx')
ax.legend()
plt.show()
```

Buatlah plot dari peng-integral-an sebenarnya dari `y = t**2`.
* Tumpangkan hasilnya dengan soal no. 1. 
* Berikan warna berbeda untuk setiap plot.
* Tambahkan legend/keterangan: `x^2`, `∫ (x^2) dx (approximated)`, dan `∫ (x^2) dx (actual)

**Solusi**

Hasil dari `∫ (x^2) dx` adalah `1/3*x^3`. Siapkan terlebih dahulu data numeriknya.

```python
y_int_actual = 1/3*t**3
```

```python
fig, ax = plt.subplots()
ax.plot(t, y,  color='r', label='x^2')
ax.plot(t, At, color='k', label='∫ (x^2) dx (approximated)' )
ax.plot(t, y_int_actual, color='b', label='∫ (x^2) dx (actual)' )
ax.legend()
plt.show()
```
